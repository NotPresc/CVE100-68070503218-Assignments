# Even and Odd
iterate for loop from number 1 to n:
if the number is even, print "[number] is even"
otherwise, print "[number] is odd:

if n = 5 ==> expected output
0 is even,
1 is odd,
2 is even,
3 is odd,
4 is even,



```python
n = 10
for i in range(1,n):
    if i%2 == 0:
        print(i, "is even")
    else:
        print(i, "is odd")
```

Multiplication
-define n (n=5)
-use a loop from 1 to 10 (the values of i is 1 to 10)
-if the result of multplication (num*i) is greater than 50 => print ("The result is large")
-otherwise, print the actual result ==> print ("Result=", result)

```python
n = 7
for i in range(1,10):
    if n*i > 50:
        print("The result is large")
    else:
        print("Reuslt =", n*i)
```

Skipping the special numbers

define n
loop numbers from 1 to n
if the number is divisible 4 ==> skip it with continue
if the number is greater than 20 ==> stop the loop with break
otherwise --> print the number

```python
n = 25
for i in range(1,n):
    if i%4 == 0:
        continue
    elif i>20:
        break
    else:
        print("Result =", i)
```

Guess the number
The secret number is 777
-Ask the user to guess until they get it right,
-If the guess is wrong, print "Try again"
-If the guess is correct print "You got it" and stop the loop

```python
secret = 777
guess = int(input("Guess the three digit lucky number from 000 to 999\n"))
while secret != guess:
    print("Try again")
    guess = int(input("Guess the three digit lucky number from 000 to 999\n"))
else:
    print("You got it")
```

Shopping Cart Checkout System

-Use intitial variable total_bill=0
-Ask user to add price and add the price to total_bill,
-If total_bill>1000, print ("limit reached, You cannnot buy anymore" and break the loop)
Otherwise, ask the user to add price

```python
total_bill = 0
add = int(input("Add price"))
while total_bill <= 1000:
    print("Add more")
    add = int(input("Add price"))
    total_bill = total_bill + add
print("limit reached, you cannnot buy anymore")
```
